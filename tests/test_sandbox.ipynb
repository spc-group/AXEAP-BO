## Test gen_sticks()

```python
stick_x, stick_y = xes_calc.gen_sticks("Ti", "1+")
print(stick_x)
print(stick_y)
```

```python
m_data = pd.read_excel("./test_files/Ti1+_default_stick.xlsx")
msx = m_data["eV"].to_numpy()
msy = m_data["Intensity"].to_numpy()
print(np.array_equal(msx, stick_x))
print(np.array_equal(msy, stick_y))
print(np.sum(msy))
print(np.sum(stick_y))
print(stick_y/np.sum(stick_y))
```

## Test simul_opt()

```python
varis = ["gpd"]
bounds = [(0.0, 100.0)]

exp_data = pd.read_excel("./test_files/conv_Mn2.xlsx", sheet_name="MnO_bkgrmv", dtype=float)
exp_x = exp_data["DataX"].to_numpy(dtype=float)
exp_y = exp_data["DataY"].to_numpy(dtype=float)

bayes_opt.simul_opt("Mn", "2+", exp_x, exp_y, varis, bounds, n_calls=10)
```
